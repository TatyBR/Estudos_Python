```python
projeto = input("Informe o nome o projeto:" )
horas_previstas = input("Informe a quantidade de horas previstas:" )
hora_trabalhada = input("Informe o valor da hora trabalhada (R$):" )
prazo = input("Informe o prazo estimado (meses):" )

calculo_orçamento = float(hora_trabalhada) * float(horas_previstas)
print(f"Orçamento da Obra: R$ {calculo_orçamento:.2f}")
```

```python
from fpdf import FPDF

pdf = FPDF()

pdf.add_page()
pdf.set_font("Arial", "B", 16)
pdf.image("template.png", x=0, y=0)

# Escrevendo as variáveis nas coordenadas corretas
pdf.text(115, 145, projeto)
pdf.text(115, 160, horas_previstas)
pdf.text(115, 175, hora_trabalhada)
pdf.text(115, 190, prazo)

# f"": Insere as casas decimais e já retorna como str
pdf.text(115, 205, f"{calculo_orçamento:.2f}")
# pdf.text(115, 205, str(calculo_orçamento))


pdf.output("Orçamento.pdf")
print("Orçamento gerado com sucesso!!!")
```
