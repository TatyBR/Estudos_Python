```python
df_xlsx['motivo_devolucao'].unique()
```

```python
df_xlsx['motivo_devolucao'] = df_xlsx['motivo_devolucao'].replace('S/ Devolu.','Sem Devolução')
```

```python
df_xlsx.tail()
```

```python
df_xlsx.info()
print("\n")
print(type(df_xlsx))
```

```python
# Acessando apenas a coluna clientes
df_xlsx['cliente']
```

```python
# Selecionando múltiplas colunas: passando uma lista com o nome das colunas
df_xlsx[['cliente','motorista']]
```

```python
# modificando o conteúdo dessas colunas (estava tudo maiúsculo)
df_xlsx['cliente'] = df_xlsx['cliente'].str.title()
df_xlsx['motorista'] = df_xlsx['motorista'].str.title()
```

```python
# Exportando um DataFrame

# Criando DataFrame apenas para BH
dados_xlsc_BH = df_xlsx.loc[df_xlsx['estado_destino'] == 'BH']

# Exportando este DataFrame como um excel, excluindo o índide do DataFrame
dados_xlsc_BH.to_excel('dados/base_dados_BH.xlsx', index=False)
```

### Realizando Cálculos com a Base de Dados

```python
df_xlsx.info()
```

```python
# Valores estatísticos de colunas numéricas!!!
df_xlsx['valor_pedido'].describe().round(2)
```

```python
# Visualizando os valores únicos de uma coluna
df_xlsx['cliente'].unique()
```

```python
# Quais os motoritas? (distintos)
df_xlsx['motorista'].unique()
```

```python
# Contando valores distintos: Qual motorista tem maior número de pedidos/ entrega?
df_xlsx['motorista'].value_counts()
```

```python
# Contando valores distintos: Qual a quantidade de pedidos por estado de destino?
df_xlsx['estado_destino'].value_counts()
```

#### Aplicando estatísticas

```python
# Valor médio do valor dos pedidos
df_xlsx['valor_pedido'].mean().round(2)
```

```python
# Mediana do valor dos pedidos
df_xlsx['valor_pedido'].median().round(2)
```

```python
# Desvio padrão do valor dos pedidos
df_xlsx['valor_pedido'].std()
```

```python
# Valor máximo e valor mínimo do valor dos pedidos
print(round(df_xlsx['valor_pedido'].max(),2))
print(round(df_xlsx['valor_pedido'].min(),2))
```

```python
# Qual a data do último pedido?
df_xlsx['dt_emissao_pedido'].max()
```

```python
# Qual a primeira data de entrega/ entrega mais antiga?
df_xlsx['dt_entrega'].min()
```

```python
# Qual motorista que realizou mais entregas?
df_xlsx['motorista'].mode()
```

#### Agrupando dados

```python
# Qual o faturamento total por cliente?
df_xlsx.groupby('cliente')['valor_pedido'].sum()
```

```python
df_xlsx.groupby('motorista')['valor_pedido'].sum()
```

```python
# Qual o valor médio de pedio por estado destino?
df_xlsx.groupby('estado_destino')['valor_pedido'].mean().round()
```

```python
# Agrupando por múltiplas colunas: por clinte e estado_destino
df_xlsx.groupby(['cliente','estado_destino'])['valor_pedido'].sum().to_frame()
```

```python
# Qual é o total de pedidos por clientes e por estado destino? (contando a coluna numero_pedido)
df_xlsx.groupby(['cliente','estado_destino'])['numero_pedido'].count().to_frame()
```

```python
# Qual é o total de pedidos por clientes e por status? (contando a coluna numero_pedido)
df_xlsx.groupby(['status', 'cliente'])['numero_pedido'].count().to_frame()
```

```python
# Qual é o total de pedidos por motorista e por status? (contando a coluna numero_pedido)
df_xlsx.groupby(['motorista', 'status'])['numero_pedido'].count().to_frame()
```
